What is Data Science?
Data Science is a way of getting useful information from data. A large amount of data is not useful by itself. We clean it, understand it and use it to make better decisions.

Main Steps of Data Science
Data Collection: Collecting data from CSV files, databases or websites.
Data Cleaning: Fixing missing or incorrect values in the data.
Data Analysis: Understanding data with the help of charts and numbers.
Model Building: Teaching machines to predict what may happen next.
Result Sharing: Sharing the results so that people can make better decisions.
Real-Life Examples
Netflix and YouTube suggest videos based on our interests.
Banks detect suspicious transactions.
Hospitals use patient data to predict diseases.
Weather departments predict rainfall.
Amazon and Flipkart predict which products may sell more
